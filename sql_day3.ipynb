```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("data/superstore.csv", encoding='latin1')
df.head()
```

```python
conn = sqlite3.connect("sales.db")
df.to_sql("sales", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT Region, SUM(Sales) as total_sales
FROM sales
GROUP BY Region
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT Category, SUM(Profit) as total_profit
FROM sales
GROUP BY Category
ORDER BY total_profit DESC
"""

print(pd.read_sql(query, conn))
```

```python
query = """
SELECT SubCategory, SUM(Profit) as total_profit
FROM sales
GROUP BY SubCategory
ORDER BY total_profit ASC
"""

print(pd.read_sql(query, conn))
```

```python
query = """
SELECT "Customer Name", SUM("Sales") as total_sales
FROM sales
GROUP BY "Customer Name"
ORDER BY total_sales DESC
LIMIT 5
"""

print(pd.read_sql(query, conn))
```

```python
query="""SELECT State as Top_states  ,Sum(Profit)as Total_Profit from sales group by State order by Total_Profit DESC limit 5"""
pd.read_sql(query,conn)
```

```python
query="""WITH CategoryTotals AS (
    SELECT 
        SubCategory, 
        SUM(Profit) AS Total_Profit, 
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY SubCategory
)
SELECT * 
FROM CategoryTotals
WHERE Total_Profit = (SELECT MIN(Total_Profit) FROM CategoryTotals)
   OR Total_Sales = (SELECT MAX(Total_Sales) FROM CategoryTotals);
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT [Customer Name], SUM(Sales) AS Total_Sales
FROM sales
GROUP BY [Customer Name]
HAVING SUM(Sales) > (
    SELECT AVG(CustomerTotal)
    FROM (
        SELECT SUM(Sales) AS CustomerTotal
        FROM sales
        GROUP BY [Customer Name]
    ) AS SubQuery
)order by Total_Sales DESC Limit 5;

"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT `Region`, SUM(`Sales`) AS Total_Sales,    
       SUM(`Profit`) AS Total_Profit ,
       (SUM(`Profit`)/SUM(`Sales`))*100 as Region_Performance from sales
GROUP BY `Region` Order By Region_Performance DESC"""
pd.read_sql(query,conn)
```
